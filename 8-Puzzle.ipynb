```python
# First, we will open the file, and then convert the numbers into a 3x3 Matrix.

textfile = open("Input8PuzzleCases.txt")
lines = textfile.readlines()
textfile.close()

puzzles = []

# Now lets clean the data and turn it into a 3x3 matrix.
for line in lines:
    line = line.strip()
    numbers = list(map(int, line.split(',')))

    # We create a row of 3 numbers, three times, creating a 3x3 matrix.
    puzzles.append([numbers[i:i+3] for i in range(0,9,3)])  

print(puzzles) # works as intended.

# We have successfully imported the text file and created a 3x3 matrix out of it.
```

```python
#The Goal State of the Matrix is:
goal_state = [[0,1,2],[3,4,5],[6,7,8]]

# We need to find the movable tile, 0.
def find_empty_tile(matrix):
    for i in range(3):
        for j in range(3):
            if matrix[i][j] == 0:
                return (i,j)
    return "Error"
matrix1 = puzzles[0] 

print(find_empty_tile(matrix1)) # works as intended.
```

```python
# Now we must find the possible moves for that 0 tile.

def find_possible_moves(matrix):
    row, col = find_empty_tile(matrix)
    legal_moves = []
    directions = [(-1, 0, "Up"), (0, 1, "Right"), (1, 0, "Down"), (0, -1, "Left")]

    for x, y, move in directions:
        new_row, new_col = row + x, col + y

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            legal_moves.append((move, new_row, new_col))

    return legal_moves


print(find_possible_moves(matrix1)) # works as intended.
```

```python
# Making a temporary copy of the matrix so we dont affect the original. 

from copy import deepcopy

def apply_move(matrix, move):
    row, col = find_empty_tile(matrix)
    new_row, new_col = move[1], move[2]

    new_matrix = deepcopy(matrix)  # Ensure we don’t modify original

    # Swap tiles
    new_matrix[row][col], new_matrix[new_row][new_col] = new_matrix[new_row][new_col], new_matrix[row][col]

    return new_matrix

legalmoves = find_possible_moves(matrix1)
print(matrix1)
print(apply_move(matrix1, legalmoves[0]))
```

```python

#implementation of function "breadthFirstSearch" 

def breadth_first_search(start, goal):

    # first a queue to store starting state and moves.
    queue = [(start,[])]
    
    visited = set() # sets are faster lookup than lists.

    while queue:
        current_state, path = queue.pop(0)

        if current_state == goal:
            return path
        if str(current_state) in visited:
            continue # skip this iteration if already visited.
        visited.add(str(current_state))

        for move in find_possible_moves(current_state):
            new_state = apply_move(current_state, move)
            queue.append((new_state, path + [move]))
    return None

# testing
if breadth_first_search(matrix1, goal_state):
    print("Solution found") # properly printing!
```

```python
def depth_limited_search(matrix, goal, depth, path, visited):
    if matrix == goal:
        return path  # No need to change anything.

    if depth == 0:
        return None  # Stop at depth limit

    visited.add((str(matrix)))  # We'll store the matrices as strings for hashing in visited set instead of a list for performance.

    for move in find_possible_moves(matrix):
        new_state = apply_move(matrix, move)

        if str(new_state) not in visited:
            result = depth_limited_search(new_state, goal, depth - 1, path + [move], visited)
            if result:
                return result

    return None  # Meaning solution at this depth or no solution at all

# Lets create a solver for this.
def iddfs_solver(start, goal, max_depth=50):
    for depth in range(max_depth):
       # print(f"Searching at depth: {depth}")  # Unnecessary line but useful for debugginng.
        result = depth_limited_search(start, goal, depth, [], set()) #I used a list first but it was super slow O(n) time, so a set is O(1) better.
 
        if result: # if we find a result, print and return.
            print("Solution found!")
            return result  

    print("No solution found within max depth")
    return None # no solution at max depth.

# testing
print(iddfs_solver(matrix1, goal_state, 50))
```

```python
import time
# lets compute our averages now.
def compute_average_performance(puzzles, test_count=1):
    bfs_steps = []
    bfs_times = []
    iddfs_steps = []
    iddfs_times = []
    
    print("Computing Averages...\n")
    
    # Selecting only `test_count` puzzles from the dataset because of the huge computation time. I think 10 should be good enough?
    selected_puzzles = puzzles[:test_count]

    for puzzle in selected_puzzles:
        # Testing BFS
        start_time = time.time()
        bfs_solution = breadth_first_search(puzzle, goal_state)
        bfs_time = time.time() - start_time

        if bfs_solution:
            bfs_steps.append(len(bfs_solution))
            bfs_times.append(bfs_time)

        # Testinng IDDFS
        start_time = time.time()
        iddfs_solution = iddfs_solver(puzzle, goal_state, 50)
        iddfs_time = time.time() - start_time

        if iddfs_solution:
            iddfs_steps.append(len(iddfs_solution))
            iddfs_times.append(iddfs_time)

    # Computing the averages
    avg_bfs_steps = sum(bfs_steps) / len(bfs_steps) if bfs_steps else 0 # need to handle cases where there is no solution for all of these.
    avg_bfs_time = sum(bfs_times) / len(bfs_times) if bfs_times else 0
    avg_iddfs_steps = sum(iddfs_steps) / len(iddfs_steps) if iddfs_steps else 0
    avg_iddfs_time = sum(iddfs_times) / len(iddfs_times) if iddfs_time else 0

    return {
        "avg_bfs_steps": avg_bfs_steps,
        "avg_bfs_time": avg_bfs_time,
        "avg_iddfs_steps": avg_iddfs_steps,
        "avg_iddfs_time": avg_iddfs_time
    }

# Run performance test on the first 10 puzzles
# compute_average_performance(puzzles, test_count=5)
```
